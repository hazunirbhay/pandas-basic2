Q1. Generate an array of 10 random integers (1-100), sort it, and print only the even indexed numbers.

```python
import random

# Generate an array of 10 random integers (1-100)
random_array = [random.randint(1, 100) for _ in range(10)]
print(f"Original array: {random_array}")

# Sort the array
sorted_array = sorted(random_array)
print(f"Sorted array: {sorted_array}")

# Print only the even-indexed numbers
even_indexed_numbers = [sorted_array[i] for i in range(len(sorted_array)) if i % 2 == 0]
print(f"Even-indexed numbers: {even_indexed_numbers}")
```

Q2. Create a line chart using python (Matlplotlib or seaborn) to visualize the monthly sales data of a company given in a dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Create a sample monthly sales dataset
np.random.seed(42) # for reproducibility
months = pd.date_range(start='2023-01-01', periods=12, freq='MS')
sales_data = np.random.randint(1000, 5000, size=12) + np.random.randint(-500, 500, size=12)
sales_data[sales_data < 0] = 0 # Ensure sales are non-negative

monthly_sales_df = pd.DataFrame({
    'Month': months,
    'Sales': sales_data
})

print("Sample Monthly Sales Data:")
display(monthly_sales_df.head())

# Create a line chart
plt.figure(figsize=(10, 6))
plt.plot(monthly_sales_df['Month'], monthly_sales_df['Sales'], marker='o', linestyle='-')
plt.title('Monthly Sales Data')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

For a dataset, write a python script that reads coloumns names, data types, and counts of unique values and saves this metadata in a.csv.

```python
import pandas as pd
import numpy as np

# --- 1. Create a sample dataset (if not provided) ---
# For demonstration, creating a new DataFrame here.
# If you have an existing DataFrame, you can replace 'df = pd.DataFrame(data)'
# with your DataFrame variable (e.g., 'df = your_existing_df').
data = {
    'ProductID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Category': ['Electronics', 'Books', 'Electronics', 'Home', 'Books', 'Electronics', 'Books', 'Home', 'Electronics', 'Books'],
    'Price': [1200.50, 25.99, 899.00, 45.75, 15.00, 1500.00, 30.00, 70.00, 600.00, 20.00],
    'Stock': [50, 200, 30, 80, 150, 20, 100, 60, 40, 120],
    'Availability': ['In Stock', 'In Stock', 'Out of Stock', 'In Stock', 'In Stock', 'In Stock', 'Out of Stock', 'In Stock', 'In Stock', 'In Stock'],
    'LastUpdate': pd.to_datetime(['2023-01-15', '2023-01-16', '2023-01-15', '2023-01-17', '2023-01-16', '2023-01-18', '2023-01-17', '2023-01-18', '2023-01-19', '2023-01-19'])
}
df = pd.DataFrame(data)

print("Sample DataFrame head:")
display(df.head())

# --- 2. Extract metadata ---
metadata = []
for col in df.columns:
    col_name = col
    col_dtype = df[col].dtype
    unique_count = df[col].nunique()
    metadata.append({
        'Column Name': col_name,
        'Data Type': col_dtype,
        'Unique Value Count': unique_count
    })

metadata_df = pd.DataFrame(metadata)

print("\nGenerated Metadata:")
display(metadata_df)

# --- 3. Save metadata to a CSV file ---
output_filename = 'dataset_metadata.csv'
metadata_df.to_csv(output_filename, index=False)

print(f"\nMetadata saved to {output_filename}")
```
